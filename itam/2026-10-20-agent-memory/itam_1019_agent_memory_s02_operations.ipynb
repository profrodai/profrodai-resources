Before an agent can change a memory, it needs a place to put
one. How we store a fact decides what we can ask later. Lucy's
pistachio paste costs 40 dollars a kilo, and in January it
goes up to 46. We try two designs in plain code. One keeps
the latest value. The other keeps every version with its dates.

```python
# 2.1 two designs: overwrite, or keep every version
print("replay of a recorded run: 11 real model calls")
print("made earlier, not live; key not used here")
flat = {"pistachio_price": 40}
flat["pistachio_price"] = 46
rows = [
    dict(key="pistachio_price", value=40, since="2025-09-01",
         until="2026-01-01", known="2025-09-01"),
    dict(key="pistachio_price", value=46, since="2026-01-01",
         until=None, known="2025-12-20")]
print("overwrite keeps:", flat)
for r in rows:
    print(r["value"], "from", r["since"], "to", r["until"],
          "| learned", r["known"])
```

Two clocks matter. The first is when a fact was true. The
second is when we learned it. Lucy heard about the new price
on December 20, ten days early. A good question names both
dates. What was the price on a day, as we knew it on another?

```python
# 2.2 ask both designs, with two dates
def price(rows, day, told="9999-12-31"):
    return [r["value"] for r in rows
            if r["known"] <= told and r["since"] <= day
            and (r["until"] is None or day < r["until"])]
print("Dec 25, the invoice:", price(rows, "2025-12-25"))
print("Jan 15, told Dec 10:", price(rows, "2026-01-15",
                                    "2025-12-10"))
print("Jan 15, told Dec 25:", price(rows, "2026-01-15",
                                    "2025-12-25"))
print("overwrite, Dec 25:", flat["pistachio_price"])
```

The trade-off is plain. Overwrite is tiny and fast, but it
answers the Dec 25 invoice with the wrong price and can't say
what Lucy knew on Dec 10. Keeping every version costs more
rows and more code, and it answers both. Pick by the
questions you expect, not by what looks tidy.

```python
# 2.3 what each design costs and what it can answer
import json
print("overwrite bytes:", len(json.dumps(flat)))
print("every version bytes:", len(json.dumps(rows)))
print("true on Dec 25, overwrite:", flat["pistachio_price"] == 40)
ok = price(rows, "2025-12-25") == [40]
print("true on Dec 25, every version:", ok)
```

Now the operations. A new fact arrives, and the agent can
add it, update an old memory, delete one that is now wrong,
or do nothing because it already knows. We keep a small
store, and every memory keeps its source. An update keeps
the old text beside the new one. A delete moves it to a trash.

```python
# 2.4 a small memory store: write, update, forget
class MemoryStore:
    def __init__(self):
        self.items, self.trash, self.n = {}, {}, 0
    def write(self, text, source):
        self.n += 1
        mid = f"m{self.n}"
        self.items[mid] = dict(text=text, source=source)
        return mid
    def update(self, mid, text, source):
        old = self.items[mid]
        self.items[mid] = dict(text=text, source=source,
                               was=old["text"])
    def forget(self, mid, why):
        self.trash[mid] = dict(self.items.pop(mid), why=why)
```

```python
# 2.5 Lucy's shop notes, each with its source
STORE = MemoryStore()
STORE.write("Dairy Direct delivers on Mondays and Thursdays.",
            "chat-1")
STORE.write("Pistachio Co delivers paste on Fridays.", "chat-1")
STORE.write("Pistachio paste costs 40 dollars a kilo.", "chat-2")
STORE.write("Cold Chain delivers cones on Wednesdays.", "chat-2")
STORE.write("Oat milk is not certified gluten free.", "chat-3")
for i, m in STORE.items.items():
    print(i, m["source"], m["text"])
```

Which operation is the hard question. Mem0 asks a model.
We ask Jev, with a typed question, and Jev gives a
probability for each choice. Jev advises, and our own code
acts. First the question, and how we find similar memories.

```python
# 2.6 the typed question, and a plain similarity search
from typesafe_sdk import Choice
QUESTION = {"decide": Choice(
    instructions=("The assistant keeps memories about Lucy's "
        "ice cream shop. A new fact has arrived. Decide what "
        "the assistant should do with its memory, given the "
        "similar memories it already holds."),
    criteria={
        "add": "Store the new fact as a new memory.",
        "update": "Revise one existing memory in place.",
        "delete": "The existing memory is now wrong. Remove it.",
        "noop": "The assistant already knows this."})}
SKIP = set("the a of is for to in it are and at on now "
           "lucy's shop".split())
def words(text):
    clean = text.lower().replace(",", "").replace(".", "")
    return set(clean.split()) - SKIP
def similar(store, fact):
    hits = []
    for i, m in store.items.items():
        a, b = words(fact), words(m["text"])
        if a & b:
            hits.append((-len(a & b) / len(a | b), i))
    return [i for score, i in sorted(hits)]
```

Here's the code that acts on Jev's answer. Below 70 percent
the function holds the write and returns a request for Lucy
to review. Above that, our code applies the choice. This
function is the only place that changes the store.

```python
# 2.7 our code applies the choice, or holds it for Lucy
GUARD, LOG, TURN, JEV = 0.70, [], [0], [0]
def apply(store, near, fact, p):
    top = max(p, key=p.get)
    if p[top] < GUARD:
        did = "hold for Lucy to review"
    elif top == "update":
        store.update(near[0], fact, f"chat-{TURN[0]}")
        did = "update " + near[0]
    elif top == "delete":
        store.forget(near[0], "contradicted")
        did = "delete " + near[0]
    elif top == "add":
        did = "add " + store.write(fact, f"chat-{TURN[0]}")
    else:
        did = "noop"
    LOG.append(f"  jev {top} {p[top]:.2f}, code: {did}")
    return did
```

Now the assistant. It's a request to the OpenAI Agents API
with one function tool, remember, written as plain JSON.
When Lucy says something worth keeping, the model calls
remember with a short fact. Our function asks Jev, runs our
code, and returns what it did.

```python
# 2.8 the agent request and our remember function
from agentsapi import lucy_session
from jevrec import ask
def remember(fact):
    """Save one short fact about the shop in long-term memory."""
    near = similar(STORE, fact)
    if not near:
        mid = STORE.write(fact, f"chat-{TURN[0]}")
        LOG.append("  nothing similar, code: add " + mid)
        return "saved as " + mid
    state = {"similar_memories": [STORE.items[i]["text"]
                                  for i in near[:3]],
             "new_fact": fact}
    JEV[0] += 1
    p = ask(f"s02-jev-{JEV[0]}", state, QUESTION)["decide"]
    return apply(STORE, near, fact, p)
REMEMBER_TOOL = {"type": "function", "name": "remember",
    "description": remember.__doc__,
    "parameters": {"type": "object",
        "properties": {"fact": {"type": "string"}},
        "required": ["fact"], "additionalProperties": False}}
RULES = ("You are the memory of Lucy's ice cream shop "
         "assistant. When Lucy tells you a fact worth keeping, "
         "call remember once with one short fact. Then answer "
         "in one short sentence.")
request = {"model": "gpt-6-astra", "instructions": RULES,
           "tools": [REMEMBER_TOOL]}
```

Six things Lucy says over a few weeks. The model decides
what to pass to remember. Jev says what to do with it, as a
probability for each choice. Then our code does it. Each
case prints what Lucy said, what Jev chose, and what ran.

```python
# 2.9 Lucy talks, the assistant remembers (cases 1 to 3)
SAYS = [("dairy", "The dairy supplier now delivers Tuesdays."),
        ("price", "Pistachio paste now costs 46 dollars a kilo."),
        ("cones", "Cold Chain delivers cones on Wednesdays."),
        ("vanilla", "Green Farm is our new vanilla supplier."),
        ("closed", "Pistachio Co has closed down for good."),
        ("oat", "Remember that our oat milk is certified now.")]
def say(case, text):
    TURN[0] += 1
    lucy_session(f"s02-agent-{case}", request, text,
                 {"remember": remember})
    print(">", text)
    print(LOG.pop() if LOG else "  the agent saved nothing")
for case, text in SAYS[:3]:
    say(case, text)
```

```python
# 2.10 cases 4 to 6
for case, text in SAYS[3:]:
    say(case, text)
```

The store now shows the effect. An update keeps the old text
beside the new one. That's provenance, and it lets us undo a
bad update. Delete works the same way. We don't erase. We
move the memory to a trash with a reason.

```python
# 2.11 the store afterwards: what changed and where it came from
from textwrap import fill
print(len(STORE.items), "memories")
for i, m in STORE.items.items():
    print(fill(f"{i} {m['source']} {m['text']}", 62,
               subsequent_indent="   "))
    if "was" in m:
        print(fill("was: " + m["was"], 62, initial_indent="   ",
                   subsequent_indent="      "))
for i, m in STORE.trash.items():
    print(i, "in trash:", m["why"])
```
