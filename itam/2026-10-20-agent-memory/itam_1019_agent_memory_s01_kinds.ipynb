Lucy runs an ice cream shop. Her assistant is an agent on the
OpenAI Agents API. This is the Agents API, with sessions that
OpenAI keeps. It is not the Agents SDK. An agent there is a
request: a model, its instructions and its tools. It has a
tool that checks a supplier and a tool that reads a day's
orders, and our code counts every call. First, the two
functions that do the work.

```python
# 1.1 the suppliers and a lookup tool
print("replay of a recorded run: 11 real model calls")
print("made earlier, not live; key not used here")
import os, textwrap
from shop import FILES, SUPPLIERS, workspace
def supplier_for(item: str) -> str:
    """Look up which supplier brings one item, and when."""
    for line in FILES["suppliers.md"].splitlines():
        if item.lower() in line.lower():
            return line.lstrip("- ")
    return "no supplier for " + item
def orders_for(day: str) -> str:
    """List every supplier order for one day of the week."""
    names = list(SUPPLIERS)
    rows = []
    for i in range(1, 41):
        late = "late" if i % 9 == 0 else "ok"
        rows.append(f"{day[:3].upper()}-{i:03d} {names[i % 3]} "
                    f"{3 + i % 7} tubs, status {late}")
    return "\n".join(rows)
print(len(SUPPLIERS), "suppliers in Lucy's files")
print(supplier_for("oat milk"))
```

The request lists each tool as a plain JSON spec: a name, a
description and the arguments it takes. The instructions say
to check a supplier before answering, unless the conversation
already shows it. The handlers dict is for replay. It maps a
tool's name to our function, so our code runs again on the
arguments the model chose.

```python
# 1.2 the request: model, instructions and tools
from agentsapi import lucy_session, MODEL
SUPPLIER_TOOL = {
    "type": "function", "name": "supplier_for",
    "description": "Look up which supplier brings one item, "
                   "and when.",
    "parameters": {
        "type": "object",
        "properties": {"item": {"type": "string"}},
        "required": ["item"], "additionalProperties": False}}
ORDERS_TOOL = {
    "type": "function", "name": "orders_for",
    "description": "List every supplier order for one day of "
                   "the week.",
    "parameters": {
        "type": "object",
        "properties": {"day": {"type": "string"}},
        "required": ["day"], "additionalProperties": False}}
RULES = ("You help Lucy run her ice cream shop. Her files "
         "are in /workspace. Use supplier_for to check a "
         "supplier before you answer, unless this "
         "conversation already shows it. Answer in one "
         "sentence.")
request = {"model": MODEL, "instructions": RULES,
           "tools": [SUPPLIER_TOOL, ORDERS_TOOL]}
HANDLERS = {"supplier_for": supplier_for,
            "orders_for": orders_for}
say = lambda s: print(textwrap.fill(s, 64))
print("model:", request["model"])
print("tools:", [t["name"] for t in request["tools"]])
```

Lucy asks twice, in two separate sessions. Each session
starts empty, so the second one pays for the same lookup
again. The runs were recorded once. This cell replays the
saved words and runs our tool again, so the counts are ours.

```python
# 1.3 two sessions, no memory between them
ASK1 = "Which supplier brings the oat milk, and on which days?"
ASK2 = "Lucy asks again: on which days does the oat milk come?"
n1 = lucy_session("s01-nomem-1", request, ASK1, HANDLERS)
n2 = lucy_session("s01-nomem-2", request, ASK2, HANDLERS)
print("separate sessions, lookups:", len(n1.tool_calls),
      len(n2.tool_calls))
say(n2.final_output)
```

Now the same two questions in one session. The first input
creates it. The second input goes to the same session, and
the session already holds the first answer.

```python
# 1.4 one session, continued with a second input
s1 = lucy_session("s01-one-1", request, ASK1, HANDLERS,
                  session="lucy")
s2 = lucy_session("s01-one-2", request, ASK2, HANDLERS,
                  session="lucy")
print("one session, lookups:", len(s1.tool_calls),
      len(s2.tool_calls))
print("same session:", s1.session_id == s2.session_id)
say(s2.final_output)
```

A bigger window doesn't change this, because the window
belongs to one request. Here's what one request carries,
counted by zone, at four characters per token. That's an
estimate, not a tokenizer.

```python
# 2.1 the context, counted by zone
zones = {
    "instructions": RULES,
    "tools": str(SUPPLIER_TOOL),
    "history": ASK1 + " " + ASK2,
    "answer": s2.final_output,
}
WINDOW = 200_000
tokens = {z: len(t) // 4 for z, t in zones.items()}
for z, n in tokens.items():
    print(f"{z:13} {n:4} tokens {n / WINDOW:6.2%} of the window")
print("total", sum(tokens.values()), "estimated tokens")
```

There is plenty of room, so a bigger window sounds like the
answer. Now a long week. The same session gets five more
inputs, one for each day, and each one reads that day's orders
with a tool. Then we ask for the first thing it checked.

```python
# 2.2 a long week in the same session
DAYS = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
for d in DAYS:
    w = lucy_session(f"s01-week-{d.lower()}", request,
                     f"Read {d}'s orders. How many are late?",
                     HANDLERS, session="lucy")
    print(f"{d:10} lookups {len(w.tool_calls)}  items now "
          f"{len(w.items)}")
```

The session kept all of it, and OpenAI says it manages context
compaction for us. We don't claim that it happened. We ask the
session a question about the very first thing, and then we
count what the saved items hold.

```python
# 2.3 does the first answer survive a long week?
q = lucy_session("s01-week-first", request,
                 "At the start, which supplier brought the "
                 "oat milk? Do not look it up again.",
                 HANDLERS, session="lucy")
say(q.final_output)
print("lookups for this question:", len(q.tool_calls))
kinds = sorted({i["type"] for i in q.items})
print("item types in the session:", kinds)
print("compaction item in the session:",
      any("compaction" in i["type"] for i in q.items))
```

So memory is a different job from a big window. Agents keep
four kinds, and each has its own lifetime. Working memory is
the turn itself: the input and the tool calls. Here's the
first turn from the one-session run.

```python
# 3.1 working memory: the turn
print("input:", ASK1)
for call in s1.tool_calls[:4]:
    print(" ", call["name"], call["args"])
print("tool calls in this turn:", len(s1.tool_calls))
```

Episodic memory is what happened, and the session keeps it.
We read the saved items back, the way a later turn would.
Each item is a message, a tool call, a tool result or a
command the sandbox ran.

```python
# 3.2 episodic memory: the session items
from collections import Counter
count = Counter(i["type"] for i in q.items)
for kind, n in sorted(count.items()):
    print(f"{kind:22} {n}")
print("items in the session:", len(q.items))
```

Semantic memory holds facts, and procedural memory holds a way
of doing something. Both are files in the sandbox. The facts
are Lucy's three files. The way of doing it is a skill: a
folder with a SKILL.md, registered as a capability directory.

```python
# 3.3 semantic and procedural memory as files
import base64
SKILL = ("---\nname: reorder\ndescription: How Lucy reorders a "
         "flavour when a tub count is low.\n---\n"
         "1. Read flavours.md and find the flavour.\n"
         "2. If it has 3 tubs or fewer, find its supplier in "
         "suppliers.md.\n"
         "3. Say what to order and which day it arrives.\n")
env = workspace()
env["files"].append({
    "type": "inline",
    "path": "/workspace/capabilities/shop/reorder/SKILL.md",
    "data": base64.b64encode(SKILL.encode()).decode()})
env["capability_directories"] = ["/workspace/capabilities/shop"]
print(sorted(f["path"].split("/workspace/")[1]
             for f in env["files"]))
```

A brand new session starts with those files and nothing else.
It has no history. We ask which flavours are low, and what to
order, so it has to read the facts and follow the skill.

```python
# 3.4 a new session that starts from the files
files_request = {"model": MODEL, "tools": [], "instructions": (
    "You help Lucy run her ice cream shop. Her files are in "
    "/workspace. Use the reorder skill. Answer in two "
    "sentences.")}
f = lucy_session("s01-files", files_request,
                 "Which flavours are low, and what do we order?",
                 {}, environment=env)
say(f.final_output)
print("commands the sandbox ran:",
      sum(i["type"] == "command_execution" for i in f.items))
```
